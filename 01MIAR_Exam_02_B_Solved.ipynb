## 02 - Señal de prueba (2 ptos)

Crea una función `test_signal(l, **kwargs)` que devuelva un array de numpy binomial que contiene sólo `1`s y `0`s arreglados de forma aleatoria y con una probabilidad de 50%/50%. La longitud del array está definida por el parámetro `l` cuyo valor por defecto es 10. Cómo parámetros opcionales puedes pasar una semilla `s` para que la función retorne consistentemente los mismos valores o una nueva probabilidad `p` con la que aparecerá el `1`.

Revisar [documentación](https://numpy.org/doc/stable/reference/random/generated/numpy.random.binomial.html).

```python
# Your code HERE
def test_signal(l=10, **kwargs):

     p = kwargs.get("p", 0.5)

     s = kwargs.get("s", None)

     if s is not None:
        np.random.seed(s)

     return np.random.binomial(1, p, l)
```

```python
# Test Use Case
output1 = test_signal()
print(f"Señal por defecto: {output1}")
assert(len(output1) == 10)

output2 = test_signal(5)
print(f"Señal de 5 valores: {output2}")
assert(len(output2) == 5)

output3 = test_signal(30, p=0)
display(f"Señal con 30 0s: {output3}")
assert(output3.sum() == 0)

output4 = test_signal(300, s=42)
output5 = test_signal(300, s=42)
assert((output4 == output5).all())
```

## 03 Extraer de texto - (2 ptos)

Adjunto encontrarás un fichero llamado **texto.txt**. Crea un conjunto (set) que contenga todas las palabras en este fichero que comiencen por la letra **e**. Para ello convierte todas las mayúsculas en minúsculas y elimina espacios, caracteres de nueva línea y los signos de puntuación **"."** y **","** antes de guardarlos en el conjunto.

```python
# Your code HERE
# Your code HERE
def extract_words_starting_with_e(filename):

    with open(filename, "r", encoding="utf-8") as f:
        texto = f.read().lower()

    texto = texto.replace("\n", " ")
    texto = texto.replace(".", "")
    texto = texto.replace(",", "")

    palabras_e = {
        palabra
        for palabra in texto.split()
        if palabra.startswith("e")
    }

    return palabras_e

```

```python
def extract_words_starting_with_e(filename):
    with open(filename, "r", encoding="utf-8") as f:
        resultado = set()

        for line in f.readlines():
            for word in line.split():
                w = word.lower()
                w = w.replace(".", "").replace(",", "")
                w = w.replace("?", "").replace("!", "")

                if w.startswith("e"):
                    resultado.add(w)

    return resultado
```

```python
# Test Use Case
filename = "texto.txt"

expected_output = {'etiam', 'eleifend', 'eu', 'enim', 'est', 'eget', 'elementum', 'ex',
                   'elit', 'egestas', 'euismod', 'efficitur', 'et', 'eros', 'erat'}

output = extract_words_starting_with_e(filename)
display(output)

assert(output == expected_output)
```

```python
import pandas as pd

# Cargar datos
df = pd.read_csv("bicis.csv")

# Convertir fechas a datetime
df["started_at"] = pd.to_datetime(df["started_at"])
df["ended_at"] = pd.to_datetime(df["ended_at"])

# Calcular duración del viaje en minutos
df["duration"] = (
    df["ended_at"] - df["started_at"]
).dt.total_seconds() / 60
```

```python
# Crear un grupo por estación de partida
station_id_groupby = df.groupby("start_station_id")
print(station_id_groupby.count().index)
```

Qué hace: Cuenta cuántas veces aparece cada end_station_id.
idxmax() devuelve el identificador con mayor frecuencia.

```python
# Porcentaje de miembros (member)
def member_percentage(df):
    counts = df['member_casual'].value_counts()
    return counts.get('member', 0) / len(df)

print(member_percentage(df))
```

```python
# Estación de llegada más frecuente
def most_common_end_station(df):
    counts = df['end_station_id'].value_counts()
    return counts.idxmax()

print(most_common_end_station(df))
```

```python
def most_common_end_station(df):
    counts = df['end_station_id'].value_counts()
    return counts.idxmax() if not counts.empty else None
print(most_common_end_station(df))
```

Qué hace: Compara cada valor de la serie con 'docked_bike'. Obtiene un array de True/False. .sum() cuenta los True

```python
# Crear una que devuelva las ocurrencias de un valor específico en una columna específica.
def count_value_in_column(value, series):
    return (series == value).sum()

count_value_in_column('docked_bike', df['rideable_type'])
print(count_value_in_column('docked_bike', df['rideable_type']))
print(df[['start_station_id','end_station_id']].head())
```

```python
# Iterar sobre el grupo y construir el nuevo dataframe utilizando las funciones definidas anteriormente.
stations = list()

for station_id, df_view in station_id_groupby:
    new_row = dict()

    new_row['station_id'] = station_id
    new_row['total_rides'] = len(df_view)
    new_row['start_member_percentage'] = member_percentage(df_view)
    new_row['most_common_end_station_id'] = most_common_end_station(df_view)
    new_row['mean_duration'] = df_view['duration'].mean()

    for bike_type in df_view['rideable_type'].unique():
        new_row[bike_type] = count_value_in_column(
            bike_type,
            df_view['rideable_type']
        )

    stations.append(new_row)

stations_df = pd.DataFrame(stations)
stations_df.head(10)
```
