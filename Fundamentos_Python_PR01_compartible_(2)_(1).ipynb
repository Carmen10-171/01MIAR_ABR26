### Bloque 2 — Aritmética y el peligro de los `float`

**Concepto:** Python sabe sumar, restar, multiplicar, dividir y calcular módulo (`%`, el resto de la división). **Pero los números decimales (`float`) NO son exactos** — la computadora los guarda en binario y a veces aparece "ruido" en los últimos decimales. Esto es origen de muchos bugs y es **directamente relevante para FRange**.

La forma estándar de evitarlo: redondear con `round(valor, decimales)`.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# Enteros: siempre exacto
print(7 + 3, 7 - 3, 7 * 3, 7 / 3, 7 // 3, 7 % 3)

print("---")

# Floats: AQUÍ aparece el ruido
print(0.1 + 0.2)            # ¿esperabas 0.3?
print(0.1 + 0.2 == 0.3)     # ¿True o False?

print("---")

# La solución: round(valor, decimales)
resultado = round(0.1 + 0.2, 2)
print(resultado, resultado == 0.3)
````

### Bloque 5 — `for i in range(n)` (recorrer por índice)

**Concepto:** a veces no quieres el valor, sino la **posición** (el índice). Para eso usas `range(n)`, que genera `0, 1, 2, ..., n-1`.

Variantes:

- `range(5)` → `0, 1, 2, 3, 4` (empieza en 0, termina antes de 5).
- `range(2, 7)` → `2, 3, 4, 5, 6`.
- `range(0, 10, 2)` → `0, 2, 4, 6, 8` (con paso 2).

**Detalle CLAVE para FRange:** `range` solo acepta **enteros**, no acepta `0.4`. Por eso FRange usa `while`, no `for range`.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
for i in range(5):          # 0..4
    print(i)
print("---")
for i in range(2, 6):       # 2..5
    print(i)
print("---")
for i in range(0, 10, 2):   # 0, 2, 4, 6, 8
    print(i)
print("---")
colores = ["rojo", "verde", "azul"]
for i in range(len(colores)):
    print(i, colores[i])
````

### Bloque 6 — `enumerate` (índice **y** valor a la vez)

**Concepto:** `enumerate(lista)` te da, en cada vuelta, **una pareja (índice, valor)**. Es la forma idiomática y más limpia de recorrer cuando necesitas ambos.

Comparación:

```python
# Forma verbosa
for i in range(len(colores)):
    print(i, colores[i])

# Forma limpia con enumerate
for i, color in enumerate(colores):
    print(i, color)
```

**Sintaxis nueva:** los **dos nombres separados por coma** (`for i, color in ...`) se llaman *desempaquetado*. Python abre la pareja en dos variables.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
frutas = ["manzana", "banana", "cereza"]
for indice, fruta in enumerate(frutas):
    print(indice, fruta)

print("---")

for i, letra in enumerate("hola"):
    print(f"posición {i}: {letra}")

print("---")

# start=N cambia desde qué número cuenta (no salta elementos)
for n, fruta in enumerate(frutas, start=1):
    print(f"{n}. {fruta}")
````

### Bloque 7 — `while` con condición (la base directa de FRange)

**Concepto:** `for` se usa cuando **sabes cuántas vueltas** vas a dar. **`while` se usa cuando NO lo sabes** — sigues iterando *mientras se cumpla una condición*.

```python
while condición:
    # algo
    # actualizar la variable de la condición  ← CRÍTICO
```

**El peligro: bucle infinito.** Si te olvidas de actualizar la variable, el `while` nunca termina. Si te pasa, interrumpe la celda (botón cuadrado ⬛).

**Por qué es clave para FRange:** `range` no acepta `float`s. Para generar `30, 30.4, 30.8, ...` necesitas un `while` con suma decimal.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# while básico
i = 0
while i < 5:
    print(i)
    i += 1

print("---")

# while con paso decimal — FRange en miniatura
valor = 0.0
while valor <= 2.0:
    print(valor)
    valor += 0.5

print("---")

# while con condición compuesta
total = 0
numero = 1
while total < 100:
    total += numero
    numero += 1
print(f"total={total}, numero={numero}")
````

### Bloque 8 — Acumular en una lista vacía (FRange ya casi sale)

**Concepto:** muchas veces necesitas **construir** una lista a partir de otros datos. El patrón es siempre el mismo:

```python
resultado = []                 # 1. Lista vacía
for x in algo:                 # 2. Recorres lo que te dan
    resultado.append(...)      # 3. Agregas algo en cada vuelta
```

**`lista.append(valor)`** agrega un elemento **al final**. Es el método clave de este bloque.

La combinación **`while` + `append`** es **literalmente la solución de FRange**.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# Acumular con for: cuadrados de 0 a 4
cuadrados = []
for i in range(5):
    cuadrados.append(i * i)
print(cuadrados)

print("---")

# Acumular con while: pares hasta 10
pares = []
n = 0
while n <= 10:
    pares.append(n)
    n += 2
print(pares)

print("---")

# FRange en miniatura: 0.0 a 2.0 con paso 0.5
resultado = []
valor = 0.0
while valor <= 2.0:
    resultado.append(round(valor, 2))
    valor += 0.5
print(resultado)
````

### Bloque 9 — `if/else` y operadores de comparación

**Concepto:** un `if` ejecuta código **solo si una condición es verdadera**. Si quieres dos caminos opuestos, usas `if/else`. Si hay más de dos, usas `elif`.

**Comparación:** `==` (igual), `!=` (distinto), `<`, `>`, `<=`, `>=`. Devuelven `True` o `False`.

**Cuidado con `=` vs `==`:** `=` asigna, `==` compara. Confundirlos es el error #1 del principiante.

**Operadores lógicos:** `and` (ambas True), `or` (al menos una True), `not` (invierte).
#### Mi ejemplo

Predice qué imprime antes de correr.

### Bloque 10 — Filtrar dentro de un bucle (base directa de Florida Cities)

**Concepto:** combinar **recorrer** + **decidir** + **acumular**. Patrón:

Predice qué imprime antes de correr.

### Bloque 11 — Comparar con el vecino (`i+1`) — base del RLE

**Concepto:** muchos problemas requieren comparar **cada elemento con el siguiente** o con **el anterior**. Por ejemplo: contar duplicados consecutivos, detectar subidas, agrupar rachas (esto es RLE).

**Cuidado con los bordes:**

- Si comparas `lista[i]` con `lista[i+1]`, no te puedes salir por la derecha. Usa `range(len(lista) - 1)`.
- Si comparas `lista[i]` con `lista[i-1]`, no te puedes salir por la izquierda. Usa `range(1, len(lista))`.

Esa resta de 1 es el famoso `n-1` que confunde al principio.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# Contar cuántas veces hay dos elementos iguales consecutivos
lista = [1, 1, 2, 3, 3, 3, 4, 5, 5]
iguales = 0
for i in range(len(lista) - 1):
    if lista[i] == lista[i+1]:
        iguales += 1
print(f"Pares consecutivos iguales: {iguales}")

print("---")

# Contar cuántas veces la secuencia BAJA
valores = [3, 5, 2, 8, 6, 4, 4, 1]
bajadas = 0
for i in range(len(valores) - 1):
    if valores[i] > valores[i+1]:
        bajadas += 1
print(f"Bajadas: {bajadas}")
````

## Nivel 6 — Recursión (el bloque crítico)

**Recursión** = una función que se llama a sí misma. Suena confuso, pero el patrón es siempre el mismo:

1. **Caso base** — la entrada más pequeña posible, que se resuelve trivialmente sin más llamadas.
2. **Caso recursivo** — reduces el problema (entrada un poco más pequeña) y llamas a la misma función con esa entrada reducida.

Sin caso base → recursión infinita → `RecursionError`.

**Necesaria para el tercer ejercicio del PR01 (RLE recursivo).**

### Bloque 17 — Recursión sobre listas — caso base + reducir

**Concepto:** función que se llama a sí misma con una entrada más pequeña hasta llegar al caso base.

Plantilla universal:

```python
def f(entrada):
    if <caso base>:
        return <valor trivial>
    # caso recursivo: procesar la "cabeza" + recurrir con "cola"
    return <algo> + f(<entrada más pequeña>)
```

Las llamadas se **apilan** hasta llegar al caso base, y luego se **desapilan** combinando los resultados.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# Suma recursiva de una lista
def suma_lista(lista):
    if len(lista) == 0:          # caso base: lista vacía
        return 0
    return lista[0] + suma_lista(lista[1:])   # primer elemento + suma del resto

print(suma_lista([1, 2, 3, 4]))   # 10

print("---")

# Factorial: n! = n * (n-1)!
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))   # 120
````

### Bloque 19 — Recursión + slicing — el camino directo a RLE

**Concepto final:** combinar todo lo anterior. En RLE recursivo:

1. **Caso base:** string vacío → devolver `""`.
2. **Caso recursivo:**
   - Tomar la primera letra (`s[0]`).
   - Contar cuántas veces se repite seguidas al inicio.
   - Devolver `letra + str(count) + rle(resto)`, donde `resto = s[count:]`.

El truco fundamental: **`s[count:]` es "el resto después de saltarse las primeras `count` letras iguales"**. Eso reduce el problema.
#### Mi ejemplo

Predice qué imprime antes de correr.

````python
# Función auxiliar: cuenta cuántas veces el char inicial se repite seguidas
def contar_iniciales(s):
    if s == "":
        return 0
    if len(s) == 1 or s[0] != s[1]:
        return 1
    return 1 + contar_iniciales(s[1:])

print(contar_iniciales("dddw"))    # 3
print(contar_iniciales("a"))        # 1
print(contar_iniciales("abc"))      # 1

print("---")

# Mini-RLE recursivo (versión con bucle while embebido para contar)
def rle(s):
    if s == "":
        return ""
    # contar cuántas letras iguales hay al inicio
    count = 1
    while count < len(s) and s[count] == s[0]:
        count += 1
    return s[0] + str(count) + rle(s[count:])

print(rle("aabbb"))                  # 'a2b3'
print(rle("dddddwwoooodddaaa"))       # 'd5w2o4d3a3'
````
