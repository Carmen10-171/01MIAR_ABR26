## Pregunta 1 – Lógica proposicional

**Enunciado:** Sean p, q proposiciones lógicas tales que p ∨ q es un enunciado contradictorio.

**Respuesta:** d) ¬p ∧ ¬q es tautológica.

**Explicación sencilla:**  
Una contradicción es algo que siempre es falso. La disyunción p ∨ q solo es falsa cuando p y q son falsas al mismo tiempo.

\[
p = F,  q = F
\]

Entonces:

\[
¬p = V,  ¬q = V
\]

Por eso:

\[
¬p ∧ ¬q = V
\]

**Idea clave:**  
p ∨ q solo falla cuando ambas proposiciones son falsas.

**Ejercicio adicional:**  
Si p ∨ q es falso, calcula el valor de verdad de p ∧ q.

**Respuesta:** d) El enunciado no se cumple para ningún x ∈ ℝ.

**Idea clave:**  
Para negar un “para todo”, basta encontrar un solo contraejemplo.

## Pregunta 3 – Cardinalidad e inyectividad

**Enunciado:** Sean A, B conjuntos tales que |A|=6^2, |B|=5 y f:A → B.

**Respuesta:** a) f no puede ser inyectiva.

**Explicación sencilla:**

\[

\]

El dominio tiene 36 elementos y el codominio solo 5.

## Pregunta 5 – Continuidad y derivabilidad

**Enunciado:**

\[
f(x)=
\begin{cases}
\ln(x^2+1), & x<0\\
x^2-\sin x, & x\ge 0
\end{cases}
\]

**Respuesta:** c) f es continua en x=0.

**Explicación sencilla:**  
Límite por la izquierda:

\[
\lim_{x→ 0^-}\ln(x^2+1)=\ln(1)=0
\]

Límite por la derecha:

\[
\lim_{x→ 0^+}(x^2-\sin x)=0
\]

Valor de la función:

\[
f(0)=0^2-\sin(0)=0
\]

Como todo coincide, la función es continua en 0.

Pero no es derivable, porque:

\[
\left(\ln(x^2+1)\right)'=\frac{2x}{x^2+1}
\]

y en 0 da 0, mientras que:

\[
(x^2-\sin x)'=2x-\cos x
\]

y en 0 da -1.

Las derivadas laterales son distintas.

**Idea clave:**  
Continuidad no siempre implica derivabilidad.

**Ejercicio adicional:**

\[

\end{cases}

## Pregunta 6 – Probabilidad de la unión

**Enunciado:** Sean A, B independientes, con P(A)=0.3 y P(B)=0.7. Calcular P(A\cup B).

**Respuesta:** 0,79

**Explicación sencilla:**  

\[
P(A\cup B)=P(A)+P(B)-P(A\cap B)
\]

Como son independientes:

\[
P(A\cap B)=P(A)P(B)=0.3· 0.7=0.21
\]

Entonces:

\[
P(A\cup B)=0.3+0.7-0.21=0.79
\]

Con coma decimal:

\[
0,79
\]

**Idea clave:**  
Para sucesos independientes: P(A\cap B)=P(A)P(B).

**Ejercicio adicional:**  
Si P(A)=0.4, P(B)=0.5 y son independientes, calcula P(A\cup B).

## Pregunta 8 – Diagonalización e invertibilidad

**Enunciado:** Sea A∈ℝ^{3\times 3} una matriz diagonalizable.

**Respuesta:** d) A es invertible si tiene 3 autovalores no nulos.

**Explicación sencilla:**  
Una matriz cuadrada es invertible si su determinante es diferente de cero.

Para una matriz diagonalizable, el determinante es el producto de sus autovalores:

\[
\det(A)=λ_1λ_2λ_3
\]

Si los tres autovalores son no nulos, el producto no es cero. Entonces A es invertible.

**Idea clave:**  
Una matriz es invertible si no tiene autovalores iguales a cero.

## Pregunta 10 – Determinantes, inversa y autovalores

**Enunciado:**  
\det(A)=-9, A∈ℝ^{3\times 3}.  
B es diagonalizable con autovalores:

\[
λ_1=-2,  λ_2=\frac{1}{2},  λ_3=3
\]

Calcular:

\[
\det(3A^{-1}B^3)
\]

**Respuesta:** 81

**Explicación sencilla:**  
Como la matriz es 3\times 3:

\[
\det(3M)=3^3\det(M)
\]

Entonces:

\[
\det(3A^{-1}B^3)=3^3\det(A^{-1})\det(B^3)
\]

\[
3^3=27
\]

\[
\det(A^{-1})=\frac{1}{\det(A)}=\frac{1}{-9}
\]

El determinante de B es el producto de sus autovalores:

\[
\det(B)=(-2)\left(\frac{1}{2}\right)(3)=-3
\]

\[
\det(B^3)=\det(B)^3=(-3)^3=-27
\]

Finalmente:

\[
27\left(\frac{-1}{9}\right)(-27)=81
\]

**Idea clave:**  
\[
\det(kA)=k^n\det(A)
\]
si A es n\times n.

**Ejercicio adicional:**

## Pregunta 13 – Puntos críticos

**Enunciado:**

\[
f(x,y)=1-\frac{1}{2}x^2+y^2
\]

**Respuesta:** d) (0,0) es un punto crítico y es punto de silla.

**Explicación sencilla:**  
Calculamos derivadas parciales:

\[
\frac{∂ f}{∂ x}=-x
\]

\[
\frac{∂ f}{∂ y}=2y
\]

Igualamos a cero:

\[
-x=0 \Rightarrow x=0
\]

\[
2y=0 \Rightarrow y=0
\]

El punto crítico es:

\[
(0,0)
\]

La función tiene un término negativo en x^2 y positivo en y^2:

\[
-\frac{1}{2}x^2+y^2
\]

Eso significa que en una dirección baja y en otra sube. Por eso es punto de silla.

**Idea clave:**  
Si sube en una dirección y baja en otra, es punto de silla.

**Ejercicio adicional:**

\[

## Pregunta 15 – Teorema de Bayes

**Enunciado:**  
En un grupo:

- 60 % tiene sarampión.
- De quienes tienen sarampión, 10 % no tienen manchas.
- De quienes no tienen sarampión, 10 % tiene manchas.

Calcular la probabilidad de tener sarampión sabiendo que tiene manchas.

**Respuesta:** 0,931

**Explicación sencilla:**  
Sea:

- S: tener sarampión.
- M: tener manchas.

Datos:

\[
P(S)=0.6
\]

Si 10 % de quienes tienen sarampión no tienen manchas, entonces:

\[
P(M|S)=0.9
\]

Además:

\[
P(S^c)=0.4
\]

\[
P(M|S^c)=0.1
\]

Queremos:

\[
P(S|M)
\]

Por Bayes:

\[
P(S|M)=\frac{P(M|S)P(S)}{P(M)}
\]

Calculamos P(M):

\[
P(M)=P(M|S)P(S)+P(M|S^c)P(S^c)
\]

\[
P(M)=0.9(0.6)+0.1(0.4)=0.54+0.04=0.58
\]

Entonces:

\[
P(S|M)=\frac{0.54}{0.58}=0.931034...
\]

Redondeando a tres decimales:

\[
0,931
\]

**Idea clave:**  
Bayes permite invertir probabilidades condicionadas.

**Ejercicio adicional:**  
El 30 % de una población tiene una enfermedad. El 80 % de los enfermos da positivo y el 10 % de los sanos da positivo. Calcula la probabilidad de estar enfermo si la prueba dio positivo.


## Pregunta 16 – Esperanza matemática

**Enunciado:**  
Se lanza un dado equilibrado.

- Gana 2 € si sale un número par mayor que 2.
- Pierde 1,5 € en otro caso.

**Respuesta:** b) -1/3

**Explicación sencilla:**  
Los números pares mayores que 2 son:

\[
4,6
\]

Hay 2 casos de 6:

\[
P(\text{ganar})=\frac{2}{6}
\]

Los demás resultados son:

\[
1,2,3,5
\]

Hay 4 casos de 6:

\[
P(\text{perder})=\frac{4}{6}
\]

La esperanza es:

\[
E=2\left(\frac{2}{6}\right)+(-1.5)\left(\frac{4}{6}\right)
\]

\[
E=\frac{4}{6}-\frac{6}{6}=-\frac{2}{6}
\]

\[
E=-\frac{1}{3}
\]

**Idea clave:**  
La esperanza es el promedio ponderado de ganancias y pérdidas.

**Ejercicio adicional:**  
Un dado paga 5 € si sale 6 y pierde 1 € en cualquier otro caso. Calcula la ganancia esperada.
