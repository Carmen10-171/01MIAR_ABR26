## Pregunta 1 — Equivalencia lógica

**Enunciado:**  
El enunciado `¬q → p` es equivalente a:

**Respuesta correcta:** **c) `p ∨ q`**

**Explicación sencilla:**  
Una implicación se puede transformar usando la regla:

```text
A → B ≡ ¬A ∨ B
```

Aquí:

```text
A = ¬q
B = p
```

Entonces:

```text
¬q → p ≡ ¬(¬q) ∨ p
¬(¬q) = q
```

Por tanto:

```text
¬q → p ≡ q ∨ p ≡ p ∨ q
```

**Ejercicio adicional para resolver:**  
Simplifica la proposición `¬p → q` y escoge su forma equivalente.

## Pregunta 2 — Cuándo una implicación es falsa

**Enunciado:**  
Sean `p`, `q` proposiciones lógicas tales que `q → p` es falsa. Entonces:

**Respuesta correcta:** **b) Siempre se cumple que `¬p ∧ q` es verdadera.**

**Explicación sencilla:**  
Una implicación `q → p` solo es falsa en un caso:

```text
q verdadera y p falsa
```

Es decir:

```text
q = verdadera
p = falsa
```

Por tanto, se cumple:

```text
¬p ∧ q
```

porque `¬p` es verdadera y `q` también es verdadera.

Ese valor no es entero, natural ni racional.

**Respuesta correcta:** **a) Necesariamente `|A| = |B|`.**

**Explicación sencilla:**  
Una función biyectiva es a la vez:

- Inyectiva: no repite imágenes.
- Sobreyectiva: cubre todo el conjunto de llegada.

Si una función entre conjuntos finitos es biyectiva, cada elemento de `A` se empareja con exactamente un elemento de `B`, sin sobrar ni faltar. Por eso ambos conjuntos tienen el mismo número de elementos.

```text
|A| = |B|
```

## Pregunta 5 — Combinaciones con repetición

**Enunciado:**  
Disponemos de pelotas de 5 colores distintos y queremos meter tres pelotas en una caja. ¿De cuántas formas distintas podemos hacerlo si admitimos pelotas del mismo color en la caja?

**Respuesta correcta:** **35**

**Explicación sencilla:**  
Tenemos:

```text
n = 5 colores
k = 3 pelotas
```

Como se admiten pelotas del mismo color y solo importa cuántas hay de cada color, usamos combinaciones con repetición:

```text
CR(n,k) = C(n + k - 1, k)
```

Sustituyendo:

```text
CR(5,3) = C(5 + 3 - 1, 3)
CR(5,3) = C(7,3)
CR(5,3) = 35
```

## Pregunta 6 — Límite, continuidad y derivabilidad

**Enunciado:**  
Sea `f : ℝ → ℝ` dada por:

```text
f(x) = x³              si x < 0
f(x) = e^(-x²) − 1    si x ≥ 0
```

Entonces:

**Respuesta correcta:** **d) `f` es derivable en `x = 0`.**

**Explicación sencilla:**  
Primero miramos el límite por la izquierda:

```text
lim x→0⁻ x³ = 0
```

Ahora el límite por la derecha:

```text
lim x→0⁺ (e^(-x²) − 1) = e⁰ − 1 = 1 − 1 = 0
```

Además:

```text
f(0) = e⁰ − 1 = 0
```

Entonces la función es continua en `0`.

Ahora miramos la derivada por ambos lados.

Para `x < 0`:

```text
f'(x) = 3x²
f'(0⁻) = 0
```

Para `x ≥ 0`:

```text
f'(x) = -2x e^(-x²)
f'(0⁺) = 0
```

Como las derivadas laterales coinciden, la función es derivable en `x = 0`.

**Ejercicio adicional para resolver:**  
Estudia si la siguiente función es continua y derivable en `x = 0`:

```text
f(x) = x²    si x < 0
f(x) = x     si x ≥ 0
```

**Explicación sencilla:**  
Una función sobreyectiva debe cubrir todos los elementos de `B`.

Pero si:

```text
|A| < |B|
```

entonces hay menos elementos en el dominio que en el codominio. No hay suficientes elementos en `A` para cubrir todos los elementos de `B`.

Por eso `f` no puede ser sobreyectiva.

## Pregunta 9 — Integral definida

**Enunciado:**  
El valor exacto de la integral definida es:

```text
∫₁ᵉ 4x ln(x) dx
```

**Respuesta correcta:** **b) `1 + e²`**

**Explicación sencilla:**  
Usamos integración por partes:

```text
u = ln(x)       → du = 1/x dx
dv = 4x dx      → v = 2x²
```

Entonces:

```text
∫ 4x ln(x) dx = 2x² ln(x) − ∫ 2x dx
```

```text
= 2x² ln(x) − x²
```

Evaluamos de `1` a `e`:

En `x = e`:

```text
2e² ln(e) − e² = 2e² − e² = e²
```

En `x = 1`:

```text
2(1)² ln(1) − 1² = 0 − 1 = -1
```

Por tanto:

```text
e² − (-1) = e² + 1
```

**Ejercicio adicional para resolver:**

## Pregunta 13 — Probabilidad total

**Enunciado:**  
Sean `A`, `B` una partición del espacio muestral `Ω`, es decir:

```text
Ω = A ∪ B
A ∩ B = ∅
```

Calcule la probabilidad del evento `C` si:

```text
P(B) = 0,4
P(C|A) = 0,8
P(C|B) = 0,2
```

**Respuesta correcta:** **0,56**

**Explicación sencilla:**  
Como `A` y `B` forman una partición:

```text
P(A) + P(B) = 1
```

Entonces:

```text
P(A) = 1 − 0,4 = 0,6
```

Usamos la fórmula de probabilidad total:

```text
P(C) = P(C|A)P(A) + P(C|B)P(B)
```

Sustituimos:

```text
P(C) = 0,8 · 0,6 + 0,2 · 0,4
P(C) = 0,48 + 0,08
P(C) = 0,56
```

**Ejercicio adicional para resolver:**

## Pregunta 14 — Puntos críticos y Hessiano

**Enunciado:**  
Sea:

```text
f(x,y) = x²(y − 1) − (y + 1)²
```

Entonces:

**Respuesta correcta:** **c) `(0,−1)` es un punto crítico de la función y además un máximo local.**

**Explicación sencilla:**  
Calculamos las derivadas parciales:

```text
fx = 2x(y − 1)
fy = x² − 2(y + 1)
```

Buscamos puntos críticos igualando a cero:

```text
2x(y − 1) = 0
x² − 2(y + 1) = 0
```

De la primera ecuación:

```text
x = 0   o   y = 1
```

Si `x = 0`:

```text
0 − 2(y + 1) = 0
```

```text
y = -1
```

Entonces aparece el punto:

```text
(0, -1)
```

Para clasificarlo usamos el Hessiano:

```text
fxx = 2(y − 1)
fyy = -2
fxy = 2x
```

En `(0,-1)`:

```text
fxx = -4
fyy = -2
fxy = 0
```

El Hessiano queda negativo definido, por tanto el punto es un **máximo local**.

**Ejercicio adicional para resolver:**  
Encuentra y clasifica los puntos críticos de:

```text
f(x,y) = -x² - y² + 4
```

## Pregunta 15 — Regla de la cadena en varias variables

**Enunciado:**  
Sean `g : ℝ → ℝ²` y `f : ℝ² → ℝ`, con:

```text
g(0) = (3,4)
g₁′(0) = 7
g₂′(0) = −1
∂f/∂x(3,4) = 1
∂f/∂y(3,4) = −3
```

Obtenga el valor de `[f ∘ g]′(0)`.

**Respuesta correcta:** **a) 10**

**Explicación sencilla:**  
Usamos la regla de la cadena:

```text
(f ∘ g)'(0) = fx(g(0)) · g₁'(0) + fy(g(0)) · g₂'(0)
```

Sustituimos los valores:

```text
(f ∘ g)'(0) = 1 · 7 + (-3) · (-1)
```

```text
= 7 + 3
= 10
```

## Pregunta 18 — Teorema de Bayes

**Enunciado:**  
En un grupo de personas:

```text
20 % tiene sarampión.
De quienes tienen sarampión, 5 % no tienen manchas.
De quienes no tienen sarampión, 5 % tiene manchas.
```

¿Cuál es la probabilidad de que un individuo tenga sarampión si tiene manchas?

**Respuesta correcta:** **0,83**

**Explicación sencilla:**  
Definimos:

```text
S = tener sarampión
M = tener manchas
```

Datos:

```text
P(S) = 0,20
P(no S) = 0,80
P(no M | S) = 0,05
P(M | S) = 0,95
P(M | no S) = 0,05
```

Queremos:

```text
P(S | M)
```

Usamos Bayes:

```text
P(S | M) = P(M | S)P(S) / [P(M | S)P(S) + P(M | no S)P(no S)]
```

Sustituimos:

```text
P(S | M) = (0,95)(0,20) / [(0,95)(0,20) + (0,05)(0,80)]
```

```text
P(S | M) = 0,19 / (0,19 + 0,04)
P(S | M) = 0,19 / 0,23
P(S | M) ≈ 0,826
```

Redondeando a dos decimales:

```text
0,83
```

**Ejercicio adicional para resolver:**

## Pregunta 20 — Normas vectoriales

**Enunciado:**  
Sea:

```text
v = (α, −1)
```

un vector de `ℝ²`. ¿Para cuántos valores de `α` se cumple?

```text
‖v‖₂² = ‖v‖₁
```

**Respuesta correcta:** **d) 3**

**Explicación sencilla:**  
Primero calculamos cada norma.

Norma 2 al cuadrado:

```text
‖v‖₂² = α² + (-1)² = α² + 1
```

Norma 1:

```text
‖v‖₁ = |α| + |-1| = |α| + 1
```

Igualamos:

```text
α² + 1 = |α| + 1
```

Restamos 1 a ambos lados:

```text
α² = |α|
```

Sea:

```text
t = |α|
```

Entonces:

```text
t² = t
```

```text
t² - t = 0
```

```text
t(t - 1) = 0
```

Por tanto:

```text
t = 0  o  t = 1
```

Si `|α| = 0`, entonces:

```text
α = 0
```

Si `|α| = 1`, entonces:

```text
α = 1 o α = -1
```

Hay 3 valores posibles:

```text
α = -1, 0, 1
```

**Ejercicio adicional para resolver:**
