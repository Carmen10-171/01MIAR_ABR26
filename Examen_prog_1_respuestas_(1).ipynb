### Pregunta 18
**La función `n_Xs(n)` devuelve un string con `n` letras `'x'`. La versión original usa un bucle `while`:**


````python
def n_Xs(n):
    k = ''
    i = 0
    while i < n:
        k = k + 'x'
        i += 1
    return k

print(n_Xs(5))  # Esperado: 'xxxxx'
print(n_Xs(0))  # Esperado: ''
print(n_Xs(3))  # Esperado: 'xxx'
````

**Escribe la expresión que debe ir en la línea del `return` para convertir el cuerpo de la función en una única sentencia:**


````python
def n_Xs(n):
    k = ''
    i = 0
    while i < n:
        k = k + 'x'
        i += 1
    return "x" * n

print(n_Xs(5))  # Esperado: 'xxxxx'
print(n_Xs(0))  # Esperado: ''
print(n_Xs(3))  # Esperado: 'xxx'
````

### Pregunta 28
**¿Qué devuelve la llamada a la función que se escribe en la línea 10?**


````python
def pmax(lista):
    maxim = lista[0]
    pos = 0
    for i in range(1, len(lista)):
        if lista[i] > maxim:
            maxim = lista[i]
            pos = i
    return pos, maxim

print(pmax([3, 5, 5, 5, 4, 1]))
````

- A) Se produce un error de sintaxis.
- B) Ninguna de las otras respuestas es correcta.
- C) `(1, 5)`
- D) `(3, 5)`
- E) `(4, 3)`


````python
# Escribe aquí la letra de tu respuesta
C) (1, 5)
````
