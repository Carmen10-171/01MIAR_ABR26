## 01 - 1.25pt
- Crear un array de NumPy de dimensiones 5x5, con números enteros aleatorios en el rango [-40, 40]
- Hacer los siguientes reemplazos en orden
    - Reemplazar todos los valores multiplos de 3 por 1
    - Reemplazar todos los valores pares positivos por 0
    - Reemplazar todos los valores pares negativos por Nan
- Averiguar cuantos elementos tienen Nan por fila

```python
# Your solution HERE !!!!
import numpy as np

# 1. Array 5x5 con enteros aleatorios en [-40, 40]
arr = np.random.randint(-40, 41, size=(5, 5)).astype(float)
print("Array original:\n", arr)

# 2. Reemplazo 1: múltiplos de 3 → 1
arr[arr % 3 == 0] = 1

# 3. Reemplazo 2: pares positivos → 0
arr[(arr > 0) & (arr % 2 == 0)] = 0

# 4. Reemplazo 3: pares negativos → NaN
arr[(arr < 0) & (arr % 2 == 0)] = np.nan

print("\nArray modificado:\n", arr)

# 5. Contar NaN por fila
nan_por_fila = np.isnan(arr).sum(axis=1)
print("\nNaNs por fila:", nan_por_fila)
```

```python
def factor_generator(n):
    small = []
    large = []
    for i in range(1, int(n**0.5) + 1):
        if n % i == 0:
            small.append(i)
            if i != n // i:
                large.append(n // i)

    for x in small:
        yield x
    for x in reversed(large):
        yield x
print(arr)
```

```python
# C. Muestra en una figura, con un tamaño de 8 de ancho y 4 de alto una gráfica de tipo puntos (scatter), la localización de los
#    incidentes en base a la longitud y latitud
df = data_aircraft

df_filtrado = df[
    (df["LATITUDE"] >= -90) & (df["LATITUDE"] <= 90) &
    (df["LONGITUDE"] >= -180) & (df["LONGITUDE"] <= 180)
]

plt.figure(figsize=(10,5))
plt.scatter(df_filtrado["LONGITUDE"], df_filtrado["LATITUDE"],
            s=5, edgecolor='black', linewidth=0.5, color='skyblue')

plt.title("Scatter plot")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.grid(alpha=0.4)
plt.show()
```

```python
# D. Muestra en una figura un conteo de incidentes sobre la columnab AIRCRAFT por año, en el gráfico debe aparecer una columna por cada
# DAMAGE_LEVEL distinto en el dataframe. La gráfica debe ser de tipo linea con diferentes subplots

df["INCIDENT_DATE"] = pd.to_datetime(df["INCIDENT_DATE"])
df["YEAR"] = df["INCIDENT_DATE"].dt.year

grouped = df.groupby(["YEAR", "AIRCRAFT", "DAMAGE_LEVEL"]).size().reset_index(name="COUNT")

damage_levels = grouped["DAMAGE_LEVEL"].unique()

fig, axes = plt.subplots(1, len(damage_levels), figsize=(18, 6))

for ax, level in zip(axes, damage_levels):
    subset = grouped[grouped["DAMAGE_LEVEL"] == level]
    for aircraft in subset["AIRCRAFT"].unique():
        aircraft_data = subset[subset["AIRCRAFT"] == aircraft]
        ax.plot(aircraft_data["YEAR"], aircraft_data["COUNT"])
    ax.set_title(f"Damage Level: {level}")
    ax.set_ylabel("Incident Count")
    ax.set_xlabel("Year")

plt.tight_layout()
plt.show()
```
