- Genera una matriz con valores aleatorios de 0 a 100 de dimensiones 16x16
- Calcula los siguentes estadísticos en la matriz anterior, por fila (el resultado de cada uno debe ser un array de dimensiones 1x16)
  - Suma de los elementos de cada fila
  - Media de los elementos de cada fila
  - Valores mínimos de cada fila
  - Valores máximos de cada fila

```python
import random

# 1. Generar matriz 16x16 con valores aleatorios entre 0 y 100
matriz = [[random.randint(0, 100) for _ in range(16)] for _ in range(16)]

# 2. Suma de los elementos de cada fila (array 1x16)
suma_filas = [sum(fila) for fila in matriz]

# 3. Media de los elementos de cada fila (array 1x16)
media_filas = [sum(fila) / len(fila) for fila in matriz]

# 4. Valores mínimos de cada fila (array 1x16)
min_filas = [min(fila) for fila in matriz]

# 5. Valores máximos de cada fila (array 1x16)
max_filas = [max(fila) for fila in matriz]

# Mostrar resultados
print("Matriz 16x16:")
for fila in matriz:
    print(fila)

print("\nSuma por fila:")
print(suma_filas)

print("\nMedia por fila:")
print(media_filas)

print("\nMínimos por fila:")
print(min_filas)

print("\nMáximos por fila:")
print(max_filas)
```

```python
import pandas as pd

def convertir_extremos_mayus(lista):
    serie = pd.Series(lista)
    return serie.apply(lambda s: s[0].upper() + s[1:-1] + s[-1].upper() if len(s) > 1 else s.upper())

palabras = ["hola", "mundo", "python", "carmen"]
resultado = convertir_extremos_mayus(palabras)
print(resultado)
```

  - Orders, representa un conjunto de pedidos, con sus fechas, su cantidad, y el cliente y vendedor asociados al mismo

```python
import pandas as pd
import matplotlib.pyplot as plt

# ---- DATASETS ----

Orders = {
    "order_id": [1,2,3,4,5,6],
    "date": ["2024-08-01","2024-08-03","2024-08-10","2024-09-01","2024-09-10","2024-08-20"],
    "amount": [1200, 800, 1500, 600, 900, 2000],
    "customer_id": [1,2,3,1,2,3],
    "salesman_id": [10,11,12,10,11,12]   # ← AHORA SÍ: 6 ELEMENTOS
}

Customers = {
    "customer_id": [1,2,3],
    "customer_name": ["Aki Tanaka","Bruce Wayne","Sakura Ito"],
    "city": ["Tokyo","Gotham","Tokyo"]
}

Salesman = {
    "salesman_id": [10,11,12],
    "salesman_name": ["Ikuzu Midoriya","Clark Kent","Peter Parker"],
    "city": ["Tokyo","Metropolis","NYC"],
    "commission": [0.10, 0.15, 0.12]
}

Cities = {
    "city": ["Tokyo","Gotham","Metropolis","NYC"]
}

# ---- CREAR DATAFRAMES ----
df_orders = pd.DataFrame(Orders)
df_customers = pd.DataFrame(Customers)
df_salesman = pd.DataFrame(Salesman)
df_cities = pd.DataFrame(Cities)

# ---- UNIR DATASETS ----
df = df_orders.merge(df_customers, on="customer_id") \
              .merge(df_salesman, on="salesman_id")

# ---- RENOMBRAR COLUMNAS PARA EVITAR city_x / city_y ----
df = df.rename(columns={
    "city_x": "customer_city",
    "city_y": "salesman_city"
})

# ---- VENDEDOR CON LA MEJOR VENTA Y CLIENTE ----
best_sale = df.loc[df["amount"].idxmax()]
print("Vendedor con la mejor venta:", best_sale["salesman_name"])
print("Cliente:", best_sale["customer_name"])

# ---- COMISIÓN POR PEDIDO ORDENADA POR FECHA ----
df["commission_amount"] = df["amount"] * df["commission"]
comisiones_por_fecha = df.sort_values("date")[["date","order_id","commission_amount"]]
print("\nComisiones por fecha:")
print(comisiones_por_fecha)

# ---- MEDIA DE PEDIDOS DE CLIENTES DE TOKYO ----
media_tokyo = df[df["customer_city"] == "Tokyo"]["amount"].mean()
print("\nMedia pedidos clientes de Tokyo:", media_tokyo)

# ---- TOTAL COMISIONES DE IKUZU MIDORIYA EN AGOSTO ----
df["date"] = pd.to_datetime(df["date"])
agosto_midoriya = df[
    (df["salesman_name"] == "Ikuzu Midoriya") &
    (df["date"].dt.month == 8)
]["commission_amount"].sum()

print("\nComisiones de Ikuzu Midoriya en agosto:", agosto_midoriya)

# ---- GRÁFICO: MEDIA DE PEDIDOS POR CIUDAD ----
media_por_ciudad = df.groupby("customer_city")["amount"].mean()

media_por_ciudad.plot(kind="bar", figsize=(8,4), title="Media de pedidos por ciudad")
plt.ylabel("Media del total del pedido")
plt.show()

# ---- GRÁFICO: LÍNEA TEMPORAL DE PEDIDOS ----
df_fechas = df.sort_values("date")

plt.figure(figsize=(10,4))
plt.plot(df_fechas["date"], df_fechas["amount"], marker="o")
plt.title("Línea temporal del total de los pedidos")
plt.xlabel("Fecha")
plt.ylabel("Total del pedido")
plt.grid(True)
plt.show()
```

## 07 - 1pt
- Escribir un generador `frange` que simule el comportamiento de la función `arange` de NumPy
- Esta función cada vez que se invoque debe devolver valores espaciados uniformemente a partir de un valor inicial. Se debe proveer a dicha función un valor inicial para la secuencia de números y un valor para el step
- Debe soportar valores númericos en coma flotante
- La función potencialmente se puede llamar un número infinito de veces
- Usando plain Python, sin ninguna librería

```python
def frange(start, step):
    """Generador que simula numpy.arange usando solo Python."""
    valor = start
    while True:          # permite llamadas infinitas
        yield valor
        valor += step     # soporta floats sin problema
g = frange(10.6, 0.8)

for _ in range(10):
    print(next(g))
```

## 01
- La serie de Fibonacci es una sucesión de números, en la cual cada número es la suma de los dos anteriores. Los dos primeros son siempre 0 y 1.
  - $F_0 = 0$
  - $F_1 = 1$
  - $F_n = F_{n-1} + F_{n-2}$
- Crear una serie de Fibonacci hasta un número *n* usando una función lambda.

```python
def fibonacci(n):
    fib = lambda a, b: a + b   # función lambda que genera el siguiente número
    serie = [0, 1]

    while serie[-1] < n:
        serie.append(fib(serie[-2], serie[-1]))

    return serie
print(fibonacci(100))
```

## 02
- Crear una función que reciba un array de NumPy y devuelva otro en el cual estén marcados como True los elementos duplicados (a partir de la segunda ocurrencia), y como False los no repetidos o las primeras ocurrencias de los duplicados.


```python
import numpy as np

def marcar_duplicados(arr):
    vistos = set()
    resultado = np.zeros_like(arr, dtype=bool)

    for i, valor in enumerate(arr):
        if valor in vistos:
            resultado[i] = True
        else:
            vistos.add(valor)

    return resultado

arr = np.array([4, 7, 4, 9, 7, 7, 2])
print(arr)
print(marcar_duplicados(arr))   # ← nombre correcto
```

- Dada una lista de elementos, crea una función que devuelva un dataframe sin los elementos duplicados


```python
import pandas as pd

def dataframe_sin_duplicados(lista):
    vistos = set()
    unicos = []

    for elemento in lista:
        if elemento not in vistos:
            vistos.add(elemento)
            unicos.append(elemento)

    return pd.DataFrame(unicos, columns=["valor"])

lista = ["a", "b", "c", "a", "c", "a", "g"]

df = dataframe_sin_duplicados(lista)
print(df)
```

# 07
- La serie de Fibonacci es una sucesión de números, en la cual cada número es la suma de los dos anteriores. Los dos primeros son siempre 0 y 1.
  - $F_0 = 0$
  - $F_1 = 1$
  - $F_n = F_{n-1} + F_{n-2}$
- Crear un generador infinito de números de Fibonacci.

```python
def fibonacci_gen():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

g = fibonacci_gen()

for _ in range(10):
    print(next(g))
```

## 08
- Crear una función que reciba una Serie de Pandas y devuelva esa misma serie, reemplazando los espacios en blanco por el carácter menos frecuente en dicha serie.

```python
import pandas as pd

def reemplazar_espacios_por_menos_frecuente(serie):
    # Convertir todo a string para evitar errores con tipos mixtos
    serie_str = serie.astype(str)

    # Unir todos los elementos para contar caracteres
    todos = "".join(serie_str)

    # Contar frecuencia de cada carácter
    frecuencia = {}
    for c in todos:
        frecuencia[c] = frecuencia.get(c, 0) + 1

    # Encontrar el carácter menos frecuente
    menos_frecuente = min(frecuencia, key=frecuencia.get)

    # Reemplazar espacios por el carácter menos frecuente
    serie_modificada = serie_str.str.replace(" ", menos_frecuente)

    return serie_modificada

s = pd.Series(["hola mundo", "python", "a b c", "Madrid", "  espacio  "])
print(s)

print("\nSerie modificada:")
print(reemplazar_espacios_por_menos_frecuente(s))
```

---
Serie.astype(str) → asegura que todos los elementos se traten como texto
  
- "".join(serie_str) → concatena todo para poder contar caracteres.

---
Se calcula la frecuencia de cada carácter con un diccionario-
- min(frecuencia, key=frecuencia.get) → obtiene el carácter menos frecuente.

- str.replace(" ", menos_frecuente) → sustituye los espacios por ese carácter.
